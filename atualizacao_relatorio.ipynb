```python
!pip -q install papermill
import papermill as pm
import os
from datetime import date
import warnings
warnings.simplefilter('ignore')
```

```python
## CONFIGURAÇÃO ##
# Define o caminho base onde estão seus notebooks
base_path = ''  # Se os notebooks estiverem na mesma pasta, use ''

# 1. Definir a Lista de Notebooks (Ordem de Execução)
notebooks_a_rodar = [
    'Compras2025.ipynb',
    'Cotacoes2025.ipynb',
    'DimSuppliers.ipynb',
    'DimProducts.ipynb'
]

# 2. Configuração de Saída Nula (Multiplataforma)
# Descarta o resultado da execução, impedindo a criação de novos arquivos .ipynb.
if sys.platform == 'win32':
    SAIDA_NULA = 'NULL'
else:
    SAIDA_NULA = '/dev/null'

## LOOP DE EXECUÇÃO SEQUENCIAL ##
print(f"--- Iniciando a Execução Sequencial ({sys.platform}) em: {date.today().strftime('%Y-%m-%d')} ---")

for notebook_in in notebooks_a_rodar:
    caminho_entrada = os.path.join(base_path, notebook_in)

    # Adiciona kernel_name='python3' para garantir o uso do kernel padrão (Boa Prática)

    print(f"\n-> Iniciando: {notebook_in}...")

    try:
        # Executa o notebook. output_path=SAIDA_NULA impede a criação de novos .ipynb.
        pm.execute_notebook(
            input_path=caminho_entrada,
            output_path=SAIDA_NULA,
            kernel_name='python3'
        )
        print(f"  Sucesso: {notebook_in} concluído.")

    except Exception as e:
        print(f"  Erro Crítico: {notebook_in} falhou!")
        print(f"  Detalhes do Erro: {e.__class__.__name__}: {e}")
        # Interrompe o processo imediatamente após um erro crítico.
        # Caso queira continuar, remova a linha 'raise'.
        raise

print("\n--- Processo Mestre Finalizado. ---")
```
